### Análise Exploratória de Dados
----------------------------------

```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
# passo 1: leitura do dataset 

path="data/heat_exchanger.db"
engine=create_engine(f"sqlite:///{path}")

df=pd.read_sql("SELECT * FROM heat_exchanger", engine)
df['timestamp']=pd.to_datetime(df['timestamp'])
df['day_index']=(df['timestamp']-df['timestamp'].min()).dt.days
df.head()
```

```python
df.info()
```

```python
#verificar dados nulo s
df.isnull().sum()
```

```python
#verifica estatisticas
df.describe()
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
plt.plot(df['day_index'], df['heat_efficiency'], marker='o', linestyle='-', color='b', label='Heat Efficiency')
plt.title('Eficiência Térmica por Dia', fontsize=14)
plt.xlabel('Day Index', fontsize=12)
plt.ylabel('Heat Efficiency', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

```python
# Parte 03: predicao de eficiencia termica
import numpy as np
y=df['heat_efficiency'].values
x=df['day_index'].values

coef=np.polyfit(x,y,1)
print(f"Coeficientes da reta:{coef}")

pred=np.polyval(coef,x)
pred
```

```python
from matplotlib import pyplot as plt


plt.Figure(figsize=(12,6))
plt.plot(df['day_index'],df['heat_efficiency'],'o', label="Dados Reais")
plt.plot(x,pred, label="Dados Calculados")
plt.xlabel('Day Index')
plt.ylabel('Heat Efficiency')
plt.legend()
plt.grid()
plt.show()
```

```python
#Calculo do r2
residuos=y-pred 
ss_res=np.sum(residuos**2)
ss_tot=np.sum((y-np.mean(y))**2)

r2=1-(ss_res/ss_tot)
print(r2)
```
